#Grievance_Classification

## Step 1: Train the ML model in Google Colab

```python

import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
import joblib
import sklearn

# Sample complaint dataset
data = {
    "complaint": [
        "No water supply in our area",
        "Water pipe is leaking",
        "There is no drinking water",
        "Water connection is damaged",
        "No water for three days",
        "Water supply is irregular",
        "Water tank is empty",
        "No water coming from the tap",

        "Power cut in our street",
        "Electricity is not available",
        "Street lights are not working",
        "Electric pole is damaged",
        "Frequent power failure",
        "Electricity connection problem",
        "No current in our house",
        "Transformer is not working",

        "Garbage is not collected",
        "Waste is lying on the road",
        "Dustbins are overflowing",
        "Street is full of garbage",
        "No waste collection service",
        "Need cleaning of public area",
        "Garbage is scattered everywhere",
        "Drain area is dirty"
    ],
    "department": [
        "Water", "Water", "Water", "Water",
        "Water", "Water", "Water", "Water",

        "Electricity", "Electricity", "Electricity",
        "Electricity", "Electricity", "Electricity",
        "Electricity", "Electricity",

        "Sanitation", "Sanitation", "Sanitation",
        "Sanitation", "Sanitation", "Sanitation",
        "Sanitation", "Sanitation"
    ]
}

df = pd.DataFrame(data)

X = df["complaint"]
y = df["department"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

model = Pipeline([
    ("tfidf", TfidfVectorizer()),
    ("classifier", LogisticRegression())
])

model.fit(X_train, y_train)

predictions = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, predictions))
print("Scikit-learn version:", sklearn.__version__)

joblib.dump(model, "grievance_model.pkl")

print("Model saved successfully!")
```
